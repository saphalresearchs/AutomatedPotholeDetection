###  To assess the road damage due to potholes to help authorities to prioritize the repairs and its cost budgeting to maintenance

```python
# To see the masked area of the one annotated validated image:
# test_image = '/kaggle/input/potvalidation/pic-20-_jpg.rf.4b34591f154f226347c9ca9dfb33b2e4.jpg'
# test_image = '/kaggle/input/testimage/Pothole-256.jpg'
test_image = '/kaggle/input/tocalcarea/Pothole-342.jpg'

# predicting the above img with the best weighted model:
results = best_model.predict(source=test_image, imgsz=640, conf=0.5)

# ####################
print(results)

#cv2.imread(results)
# win_name = 'Detected Pothole'
# cv2.imshow(win_name, results)
# waits for user to press any key
# (this is necessary to avoid Python kernel form crashing)
# cv2.waitKey(0)

# closing all open windows
# cv2.destroyAllWindows()
# ####################

# To view the predicted img:
pred_img = results[0].plot()
pred_img_rgb = cv2.cvtColor(pred_img, cv2.COLOR_BGR2RGB)

# Get the count of masked plots to display:
# (1 predicated image + mim 1 masked img)
num_masks = len(results[0].masks.data) if results[0].masks is not None else 0
print(num_masks)
num_plots = 1 + num_masks

# Initialize the plotting with matplotlib:
fig, axes = plt.subplots(1, num_plots, figsize = (15,5))

# Display the predicted img:
axes[0].imshow(pred_img_rgb)
axes[0].set_title('Predicted Image')
axes[0].axis('off')

# If multiple masks are there, iterate each one and display all masked imges:
if results[0].masks is not None:
    # To get masks as numpy array
    masks = results[0].masks.data.cpu().numpy()
    for i, mask in enumerate(masks):
        # Make sure mask value will be binary
        # if value > 0 set it to 255 else 0
        bin_mask = (mask > 0).astype(np.uint8) * 255
        axes[i+1].imshow(bin_mask, cmap = 'gray')
        axes[i+1].set_title(f'Segmented Mask {i+1}')
        axes[i+1].axis('off')

# Assign layout and print all plots:
plt.tight_layout()
plt.show()
```

### To Caluclate Total Damaged Area of Potholes and Total Percentage of Road Damage due to These Potholes

```python
# Variable to hold individual area and total area:
total_mask_area = 0
list_area = []

# display the masked areas of potholes in a row:
fig, axes = plt.subplots(1, len(masks), figsize = (12,8))
print(axes)

# Perform calculations of areas if masks are available:
if results[0].masks is not None:
    # To get masks as numpy array
    masks = results[0].masks.data.cpu().numpy()
    print(len(masks))

    # To calculate total number of pixels in the mask:
    area_img = masks.shape[1] * masks.shape[2]
    print(area_img)

    for i, mask in enumerate(masks):
        # Make sure mask value will be binary
        # convert mask into binary
        bin_mask = (mask > 0).astype(np.uint8) * 255
        # To convert binary mask to colour
        coloured_mask = cv2.cvtColor(bin_mask, cv2.COLOR_GRAY2BGR)
        # To find contours in binary mask:
        contours, _ = cv2.findContours(bin_mask, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE) 
        # To get first contour:
        contour = contours[0]
        # Get the area of pothole through its contour
        area_contour = cv2.contourArea(contour)
        # Add the contour area to list_area array:
        list_area.append(area_contour)
        # Draw the contour on the mask
        cv2.drawContours(coloured_mask, [contour], -1, (0, 255, 0), 3)  

        # Draw green around this mask
        if (len(masks) > 1):
            axes[i+1].imshow(coloured_mask, cmap = 'gray')
            axes[i+1].set_title(f'Masked Pothole {i+1}')
            axes[i+1].axis('off')
        else:
            axes.imshow(coloured_mask, cmap = 'gray')
            axes.set_title(f'Masked Pothole {i+1}')
            axes.axis('off')

# Assign layout and print all plots:
plt.tight_layout()
plt.show()

# Calculate and print area of each pothole after images display:
for i, area in enumerate(list_area):
    print(f'Area of Pothole {i+1}: {area} pixels')
    total_mask_area += area

print('=' * 55)

# Calculate and print total damaged area and percentage of damage in the road:
print(f'Total Area of Damage by the Potholes in Predicted Image: {total_mask_area} pixels')
print(f'Total area of Image: {area_img} pixels')
print(f'Total Percentage of Damage in the Road: {(total_mask_area / area_img) * 100:.2f} %')
```

### To assess the potholes and its damage area & percentage in the road of a unseen/non-trained video:

```python
best_model = YOLO('/kaggle/working/runs/segment/train/weights/best.pt')

# Define the video path:
# test_video = '/kaggle/input/potvideorural/ruralRoad_potHoles.mp4'
# test_video = '/kaggle/input/cityroadpotholes/cityRoad_potHoles-side.mp4'
test_video = '/kaggle/input/testpotvideo/sample_video.mp4'

# Styles for Annotations Label Display 
font = cv2.FONT_HERSHEY_SIMPLEX
font_size = 1
text_pos = (40, 80)
font_colour = (255,255,255) # While coloured text
bkg_colour = (0,0,255)

# Define and Initialize var to add and remove frame by frame by appending last 10 percentage damaged area upon iteration below:
damage_deque = deque(maxlen = 10)

# To open the video:
vid = cv2.VideoCapture(test_video)
# Define the codec and create VideoWriter obj:
fourcc = cv2.VideoWriter_fourcc(*'XVID') # This used for .avi video
# To save a video (cv.VideoWriter(filename, fourcc, fps, frameSize))
output = cv2.VideoWriter('pot_mud_road.avi', fourcc, 20.0, (int(vid.get(3)), int(vid.get(4))))

# Read the whole video
while vid.isOpened():
    # Capture frame by frame
    ret, frame = vid.read()
    if ret:
        # Perform inference on the frame:
        results = best_model.predict(source=frame, imgsz=640, conf=0.25)
        processed_frm = results[0].plot(boxes=False)
        # Define and Initialize var:
        percentage_damage = 0

        # If masks are present - calculate total damage and percentage:
        if results[0].masks is not None:
            total_area = 0
            # To get masks as numpy array
            masks = results[0].masks.data.cpu().numpy()
            # For total number of pixels in the imges in frame
            area_img = frame.shape[0] * frame.shape[1]
            for mask in masks:
                bin_mask = (mask > 0).astype(np.uint8) * 255
                contour, _ = cv2.findContours(bin_mask, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)
                total_area += cv2.contourArea(contour[0])

            percentage_damage = (total_area / area_img) * 100

        # Calculate and update percentage of damage:
        damage_deque.append(percentage_damage)
        per_percentage_damage = sum(damage_deque) / len(damage_deque)

        # Draw thick line for text display:
        cv2.line(processed_frm, (text_pos[0], text_pos[1] - 10),
                 (text_pos[0] + 350, text_pos[1] - 10), bkg_colour, 40)
        # Annotate the frame with text of damaged area:
        cv2.putText(processed_frm, f'Total Damage in Road: {per_percentage_damage:.2f}', text_pos, font, font_size, font_colour, 2, cv2.LINE_AA)
        # cv2.putText(processed_frm, f'Total Damage in Road: {per_percentage_damage:.2f}%', text_pos, font, font_size, font_colour, 2, cv2.LINE_AA)
        # Write the processed frame to output video:
        output.write(processed_frm)
        # Uncomment the following 3 lines if running this code on a local machine to view the real-time processing results
        # cv2.imshow('Road Damage Assessment', processed_frame) # Display the processed frame
        # if cv2.waitKey(1) & 0xFF == ord('q'): # Press Q on keyboard to exit the loop
        #     break 
    else:
        break

# Release the video capture and video write objects
vid.release()
output.release()

# Close all the frames
# cv2.destroyAllWindows()
```
